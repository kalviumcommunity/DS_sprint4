# Understanding Notebook Cells: Code vs Markdown

This notebook demonstrates the proper use of Code cells and Markdown cells in Jupyter Notebook. We learn how to structure notebooks for clarity, readability, and professional presentation.

## 1. What are Notebook Cells?

A Jupyter Notebook is made up of individual cells. Each cell can be one of two types:

- **Code Cells**: Execute Python and return output
- **Markdown Cells**: Display formatted text (explanations, headings, documentation)

Professional notebooks combine both to create living documents: code that runs + narrative that explains.

## 2. Code Cells: Execution and Output

This section demonstrates Code cells. Notice how each cell contains Python code that runs when executed.

```python
# This is a Code cell - Python code runs here
print("Welcome to Jupyter Notebooks!")
x = 10
y = 5
result = x + y
print(f"Addition: {x} + {y} = {result}")
```

### Interpreting the Output

The Code cell above:
1. Prints a welcome message
2. Creates two variables (x=10, y=5)
3. Adds them together
4. Prints the result

Notice how Markdown can explain what Code does, and why it matters.

## 3. Working with Python Libraries

Let's import common data science libraries and demonstrate their use.

```python
# Import essential libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
print(f"NumPy version: {np.__version__}")
print(f"Pandas version: {pd.__version__}")
```

### Why Import Libraries?

Pandas and NumPy are essential for data science:
- **Pandas**: For working with tabular data (DataFrames)
- **NumPy**: For numerical operations and arrays

Always import at the beginning of your notebook so all cells have access to these tools.

## 4. Creating and Manipulating Data

Here we create a simple dataset to demonstrate data manipulation in Code cells.

```python
# Create a simple dataset using a dictionary
data = {
    'District': ['District A', 'District B', 'District C', 'District D'],
    'Rainfall_mm': [800, 1200, 950, 750],
    'Yield_tons': [45, 62, 53, 40]
}

# Convert to Pandas DataFrame
df = pd.DataFrame(data)
print("Dataset created:")
print(df)
```

### Understanding the Dataset

We created a simple dataset with 4 districts:
- **District**: Name of the district
- **Rainfall_mm**: Annual rainfall in millimeters
- **Yield_tons**: Crop yield in tons

This demonstrates how Code cells can create data, and Markdown explains what that data represents.

## 5. Basic Statistical Analysis

Let's compute some statistics to understand the data better.

```python
# Calculate statistics
mean_rainfall = df['Rainfall_mm'].mean()
mean_yield = df['Yield_tons'].mean()
max_yield = df['Yield_tons'].max()
min_yield = df['Yield_tons'].min()

print(f"Average Rainfall: {mean_rainfall:.1f} mm")
print(f"Average Yield: {mean_yield:.1f} tons")
print(f"Maximum Yield: {max_yield} tons")
print(f"Minimum Yield: {min_yield} tons")
```

### What the Statistics Tell Us

Key observations from the data:

- **Average Rainfall**: {value} mm (relatively moderate)
- **Average Yield**: {value} tons (baseline productivity)
- **Yield Range**: {min}-{max} tons (variation between districts)

This analysis demonstrates the importance of separating:
1. **Code** (calculation logic)
2. **Markdown** (interpretation of results)

## 7. Best Practices Summary

### Use Code Cells For:
- ✅ Importing libraries
- ✅ Loading and processing data
- ✅ Performing calculations
- ✅ Creating visualizations
- ✅ Testing hypotheses

### Use Markdown Cells For:
- ✅ Titles and section headings
- ✅ Explaining what code does
- ✅ Interpreting results
- ✅ Documenting assumptions
- ✅ Providing context and reasoning

## 8. The Professional Notebook Structure

Every professional notebook follows this pattern:

**[MARKDOWN]** → Title and purpose  
**[MARKDOWN]** → Section introduction  
**[CODE]** → Implementation  
**[MARKDOWN]** → Result interpretation  
**[MARKDOWN]** → Next section  
**[CODE]** → Implementation  
**[MARKDOWN]** → Result interpretation  

This creates a readable document where:
- Code shows WHAT you did
- Markdown explains WHY and SO WHAT
