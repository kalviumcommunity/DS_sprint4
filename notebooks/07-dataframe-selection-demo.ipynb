# Selecting Rows and Columns Using Indexing and Slicing

This notebook demonstrates clear and intentional data selection from Pandas DataFrames using various indexing and slicing methods.

## Part 1: Loading DataFrames

Before performing any selections, we load data into Pandas DataFrames from CSV files.

```python
import pandas as pd

# Load employee data
employees_df = pd.read_csv('../employees.csv')

# Load sales data
sales_df = pd.read_csv('../sales.csv')

# Display both DataFrames
print("=" * 50)
print("EMPLOYEES DATAFRAME")
print("=" * 50)
print(employees_df)

print("\n" + "=" * 50)
print("SALES DATAFRAME")
print("=" * 50)
print(sales_df)
```

## Part 2: Selecting Columns by Name

There are multiple ways to select columns in Pandas. This section demonstrates each method clearly.

### 2.1: Selecting a Single Column

Use bracket notation with the column name to select a single column. This returns a **Series** (1D data).

```python
# Method 1: Using square brackets with column name (returns Series)
department_series = employees_df['Department']

print("Single Column Selection (Method 1 - Using square brackets)")
print("Type:", type(department_series))
print(department_series)

print("\n" + "="*50 + "\n")

# Method 2: Using dot notation (only works with column names without spaces)
salary_series = employees_df.Salary

print("Single Column Selection (Method 2 - Using dot notation)")
print("Type:", type(salary_series))
print(salary_series)
```

### 2.2: Selecting Multiple Columns

Use a list of column names to select multiple columns. This returns a **DataFrame** (2D data).

```python
# Select multiple columns using a list
selected_columns = ['Name', 'Department', 'Salary']
employee_subset = employees_df[selected_columns]

print("Multiple Column Selection")
print("Columns selected:", selected_columns)
print("Type:", type(employee_subset))
print(employee_subset)

print("\n" + "="*50 + "\n")

# Different order - columns are selected in the order you specify
reordered_columns = ['Salary', 'Name', 'Department']
reordered_subset = employees_df[reordered_columns]

print("Multiple Column Selection (Different Order)")
print("Columns selected:", reordered_columns)
print(reordered_subset)
```

## Part 3: Selecting Rows Using Indexing

Pandas provides different methods for selecting rows. Understanding the difference between positional and label-based indexing is crucial.

### 3.1: Understanding Index vs Position

By default, Pandas uses numeric position (0, 1, 2...) as the index, but indices can be explicitly set to custom values.

```python
# Display the DataFrame with its default index
print("DataFrame with Default Numeric Index")
print(employees_df)

print("\nIndex values:", employees_df.index.tolist())
print("Columns:", employees_df.columns.tolist())

print("\n" + "="*50 + "\n")

# Create a DataFrame with custom index
custom_indexed_df = employees_df.set_index('EmployeeID')
print("DataFrame with Custom Index (EmployeeID)")
print(custom_indexed_df)
print("Index values:", custom_indexed_df.index.tolist())
```

### 3.2: Position-Based Indexing with `.iloc[]`

Use `.iloc[]` for **integer location** - selects by row number (0, 1, 2, etc.), regardless of the index.

```python
# Select a single row by position (row 0)
first_employee = employees_df.iloc[0]

print("Position-Based Selection: First Row (Position 0)")
print("Type:", type(first_employee))
print(first_employee)

print("\n" + "="*50 + "\n")

# Select specific row by position
third_employee = employees_df.iloc[2]

print("Position-Based Selection: Third Row (Position 2)")
print(third_employee)

print("\n" + "="*50 + "\n")

# Select last row using negative indexing
last_employee = employees_df.iloc[-1]

print("Position-Based Selection: Last Row (Position -1)")
print(last_employee)
```

### 3.3: Label-Based Indexing with `.loc[]`

Use `.loc[]` for **label-based** selection - selects by the actual index values (works with custom indices).

```python
# With default numeric index, .loc uses the index numbers
row_at_index_1 = employees_df.loc[1]

print("Label-Based Selection: Row with Index Label 1")
print("Type:", type(row_at_index_1))
print(row_at_index_1)

print("\n" + "="*50 + "\n")

# With custom index
employee_1002 = custom_indexed_df.loc[1002]  # Using EmployeeID

print("Label-Based Selection: Row with EmployeeID 1002 (Custom Index)")
print(employee_1002)

print("\n" + "="*50 + "\n")

# IMPORTANT: .loc with numeric index works differently than iloc with positive numbers
print("Important Distinction: .loc[2] vs .iloc[2] with Default Index")
print("\nemployees_df.loc[2] (label-based, selects row with index=2):")
print(employees_df.loc[2])
print("\nemployees_df.iloc[2] (position-based, selects row at position 2):")
print(employees_df.iloc[2])
```

## Part 4: Slicing Rows

Slicing allows you to select a range of rows. The behavior differs slightly between `.iloc[]` and `.loc[]`.

### 4.1: Position-Based Slicing with `.iloc[]`

With `.iloc[]`, slicing uses standard Python slicing (the upper bound is exclusive).

```python
# Slice rows 0 through 2 (positions 0, 1, 2)
rows_0_to_2 = employees_df.iloc[0:3]  # Note: 3 is exclusive, so this gets positions 0, 1, 2

print("Position-Based Slicing: Rows at Positions 0-2 (using .iloc[0:3])")
print(rows_0_to_2)

print("\n" + "="*50 + "\n")

# Slice rows 1 through 3
rows_1_to_3 = employees_df.iloc[1:4]

print("Position-Based Slicing: Rows at Positions 1-3 (using .iloc[1:4])")
print(rows_1_to_3)

print("\n" + "="*50 + "\n")

# Slice from middle to end
rows_from_2 = employees_df.iloc[2:]

print("Position-Based Slicing: From Position 2 to End (using .iloc[2:])")
print(rows_from_2)
```

### 4.2: Label-Based Slicing with `.loc[]`

With `.loc[]`, slicing is inclusive on both ends.

```python
# Slice rows with labels 1 through 3 (inclusive on both ends)
rows_label_1_to_3 = employees_df.loc[1:3]  # Both 1 and 3 are included

print("Label-Based Slicing: Rows with Index Labels 1-3 (using .loc[1:3])")
print(rows_label_1_to_3)

print("\n" + "="*50 + "\n")

# Slice with custom index
range_from_1002_to_1004 = custom_indexed_df.loc[1002:1004]

print("Label-Based Slicing: Custom Index from 1002-1004 (inclusive)")
print(range_from_1002_to_1004)
```

## Part 5: Selecting Specific Rows and Columns Together

This is one of the most important operations: combining row and column selection for precise data extraction.

### 5.1: Combined Selection with `.iloc[]` (Position-Based)

Use syntax: `.iloc[row_positions, column_positions]`

```python
print("DataFrame columns (for reference):")
for i, col in enumerate(employees_df.columns):
    print(f"  Position {i}: {col}")

print("\n" + "="*50 + "\n")

# Select rows 0-2, columns 1-2 (Name and Department)
subset_1 = employees_df.iloc[0:3, 1:3]

print("Combined Selection: Rows 0-2, Columns 1-2")
print("(Positions: rows [0:3], columns [1:3])")
print(subset_1)

print("\n" + "="*50 + "\n")

# Select rows 1 and 3, all columns
subset_2 = employees_df.iloc[[1, 3], :]

print("Combined Selection: Rows at Positions 1 and 3, All Columns")
print("(Positions: rows [1, 3], columns [:])")
print(subset_2)

print("\n" + "="*50 + "\n")

# Select first and last row, Name and Salary columns
subset_3 = employees_df.iloc[[0, -1], [1, 3]]

print("Combined Selection: First and Last Row, Name and Salary Columns")
print("(Positions: rows [0, -1], columns [1, 3])")
print(subset_3)
```

### 5.2: Combined Selection with `.loc[]` (Label-Based)

Use syntax: `.loc[row_labels, column_names]`

```python
# Select rows with index 1 and 3, columns Name and Salary
subset_4 = employees_df.loc[[1, 3], ['Name', 'Salary']]

print("Combined Selection: Index Labels 1 and 3, Name and Salary Columns")
print("(Labels: rows [1, 3], columns ['Name', 'Salary'])")
print(subset_4)

print("\n" + "="*50 + "\n")

# Select range of rows, specific columns
subset_5 = employees_df.loc[0:2, ['Department', 'Salary']]

print("Combined Selection: Index Range 0-2, Department and Salary")
print("(Labels: rows [0:2], columns ['Department', 'Salary'])")
print(subset_5)

print("\n" + "="*50 + "\n")

# With custom index
subset_6 = custom_indexed_df.loc[1002:1004, ['Name', 'Department']]

print("Combined Selection: Custom Index 1002-1004, Name and Department")
print("(Labels: rows [1002:1004], columns ['Name', 'Department'])")
print(subset_6)
```

### 5.3: Combined Selection with Mixed Methods

You can also use conditional statements to select rows based on column values.

```python
# Select rows where Department is 'Engineering', show Name and Salary
engineering_staff = employees_df[employees_df['Department'] == 'Engineering'][['Name', 'Salary']]

print("Conditional Selection: Engineering Department, Name and Salary")
print(engineering_staff)

print("\n" + "="*50 + "\n")

# Select rows where Salary > 60000, show Name and Salary
high_salaries = employees_df[employees_df['Salary'] > 60000][['Name', 'Salary']]

print("Conditional Selection: Salary > 60000, Name and Salary")
print(high_salaries)
```

## Part 6: Real-World Example with Sales Data

Let's apply these selection techniques to the sales DataFrame.

```python
# Display sales data
print("Sales DataFrame:")
print(sales_df)

print("\n" + "="*50 + "\n")

# Example 1: Select Product and Total columns
product_sales = sales_df[['Product', 'Total']]
print("Example 1: Product and Total Columns")
print(product_sales)

print("\n" + "="*50 + "\n")

# Example 2: Select first 3 rows
first_three_sales = sales_df.iloc[0:3]
print("Example 2: First 3 Rows (Position 0-2)")
print(first_three_sales)

print("\n" + "="*50 + "\n")

# Example 3: Select rows where Quantity > 10
high_quantity = sales_df[sales_df['Quantity'] > 10][['Date', 'Product', 'Quantity', 'Total']]
print("Example 3: Quantity > 10, Relevant Columns")
print(high_quantity)

print("\n" + "="*50 + "\n")

# Example 4: Select rows 1-3, columns Date, Product, Total
specific_subset = sales_df.iloc[1:4, [0, 1, 4]]
print("Example 4: Rows 1-3 (Position), Date, Product, Total Columns")
print(specific_subset)
```

## Part 7: Common Pitfalls and How to Avoid Them

Understanding these mistakes helps prevent selecting the wrong subset of data.

```python
print("SCENARIO: Analysis on Wrong Data Subset\n")
print("="*50 + "\n")

# MISTAKE 1: Using .loc with numeric index, thinking it works like .iloc
print("MISTAKE 1: Confusing .loc[] and .iloc[] with numeric indices")
print("\nData:")
print(employees_df)

print("\nIntended: Get row at position 2 (Charlie)")
print("\nUsing .loc[2] (label-based):")
print(employees_df.loc[2])
print("✓ Happens to work because index == position")

print("\nBut after resetting index, this breaks:")
reset_df = employees_df.reset_index(drop=True)
reset_df.index = [10, 20, 30, 40, 50]  # Custom index
print("\nDataFrame with custom index:")
print(reset_df)

print("\nUsing .loc[2] now gives wrong result:")
print(reset_df.loc[2])  # This will fail!
print("\n(IndexError because index 2 doesn't exist)")

print("\n" + "="*50 + "\n")

# MISTAKE 2: Off-by-one errors with slicing
print("MISTAKE 2: Off-by-one errors in slicing")
print("\nIntended: Get first 3 rows (positions 0, 1, 2)")
print("\nWrong: employees_df.iloc[0:3] actually gets positions 0, 1, 2")
print(employees_df.iloc[0:3])
print("NOTE: The upper bound (3) is EXCLUSIVE in .iloc[]")

print("\n" + "="*50 + "\n")

# MISTAKE 3: Forgetting inclusive behavior of .loc
print("MISTAKE 3: Forgetting that .loc[] is INCLUSIVE on both ends")
print("\nIntended: Get rows with indices 0 and 1")
print("\nUsing .loc[0:1] includes BOTH 0 AND 1:")
print(employees_df.loc[0:1])
print("(Not just index 0)")

print("\n" + "="*50 + "\n")

# SOLUTION: Always verify your selection
print("SOLUTION: Always verify your selection")
print("\n1. Check the shape of selection")
selected = employees_df.iloc[0:3, 1:3]
print(f"   Shape: {selected.shape}  (expecting 3 rows, 2 columns)")

print("\n2. Display the selected data")
print(selected)

print("\n3. Verify column names")
print(f"   Columns: {selected.columns.tolist()}")

print("\n4. Check index values")
print(f"   Index: {selected.index.tolist()}")
```

## Summary: Key Takeaways

### Column Selection
- **Single column**: `df['column_name']` returns a Series
- **Multiple columns**: `df[['col1', 'col2']]` returns a DataFrame

### Row Selection
- **Position-based** (`.iloc[]`): Uses integer positions (0, 1, 2...)
- **Label-based** (`.loc[]`): Uses index labels

### Slicing Differences
- `.iloc[start:end]`: End is **EXCLUSIVE** (Python-style)
- `.loc[start:end]`: End is **INCLUSIVE** (label-based)

### Combined Selection
- `.iloc[rows, columns]`: Both use positions
- `.loc[rows, columns]`: Both use labels
- `.loc[rows, ['col_names']]`: Mix labels with column names

### Best Practices
1. **Be explicit**: Use `.iloc[]` or `.loc[]` instead of relying on mixed syntax
2. **Verify selections**: Always print/check your selected data
3. **Use meaningful indices**: Set index to meaningful columns when helpful
4. **Document your intent**: Add comments explaining what data you're selecting
5. **Check intermediate results**: Verify data after each selection step
