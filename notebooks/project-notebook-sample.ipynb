# Jupyter Notebook Setup & Workspace Awareness

This notebook demonstrates correct Jupyter workspace setup and how to work with file paths in a data science project.

## 1. Understanding Your Working Directory

The working directory is the base location from which Jupyter looks for files.

```python
import os

# Get current working directory
working_dir = os.getcwd()
print(f"Current Working Directory: {working_dir}")

# This is where Jupyter looks for files by default
# This should be your project root: /c:/sprint-4-ds-lifecycle
```

## 2. Listing Files in Your Project

```python
# List all files in current directory
files_in_root = os.listdir('.')
print("Files in project root:")
for item in files_in_root:
    if os.path.isdir(item):
        print(f"  📁 {item}/")
    else:
        print(f"  📄 {item}")
```

## 3. Navigating to Different Folders

```python
# List files in 'data' folder using relative path
data_folder = './data'

if os.path.exists(data_folder):
    print(f"Contents of {data_folder}:")
    data_files = os.listdir(data_folder)
    for file in data_files:
        print(f"  📄 {file}")
else:
    print(f"Folder {data_folder} does not exist (but will exist when data is added)")
```

## 4. Working with File Paths - Correct Way

```python
# ✅ CORRECT: Using relative paths with os.path.join()
# This works on Windows, Mac, and Linux

data_file = os.path.join('data', 'crop_data.csv')
print(f"Correct path to data file: {data_file}")

# This is what Jupyter will look for if you try to load the file
print(f"Full path that will be used: {os.path.abspath(data_file)}")
```

## 5. Common File Path Mistakes

```python
print("❌ WRONG WAYS - DO NOT DO THIS:\n")

wrong_path_1 = "C:/Users/YourName/Desktop/crop_data.csv"
print(f"1. Hardcoded path: {wrong_path_1}")
print("   Problem: Won't work on other computers\n")

wrong_path_2 = "crop_data.csv"
print(f"2. No folder specified: {wrong_path_2}")
print("   Problem: Assumes file is in working directory\n")

wrong_path_3 = "../data/crop_data.csv"
print(f"3. Relative to notebook location: {wrong_path_3}")
print("   Problem: Breaks if notebook location changes\n")

print("\n✅ CORRECT WAY:\n")
correct_path = os.path.join('data', 'crop_data.csv')
print(f"Use os.path.join(): {correct_path}")
print("   Benefit: Works on all operating systems")
```
