# Kernel Management Demonstration

This notebook demonstrates how to run cells intentionally, interrupt long-running code, and restart the kernel cleanly.

## Key Concepts
- **Kernel State**: Variables persist across cells until restart
- **Execution Order**: Cells should run top-to-bottom
- **Interruption**: Stop a running cell without losing variables
- **Restart**: Clear all state and start fresh
- **Reproducibility**: Same results when running "Run All Cells"

## Part 1: Setup and Initial State

Run the cells below in order. Observe how each cell builds on the previous one.

```python
# Cell 1: Imports and configuration
import time
import random

print("✓ Cell 1 Complete: Libraries imported")

# Configuration variables
PROJECT_NAME = "Kernel Demo"
VERSION = 1.0
print(f"Project: {PROJECT_NAME} v{VERSION}")
```

```python
# Cell 2: Create data
# This cell depends on Cell 1 (imports are available)

dataset = {
    'batch_1': [10, 20, 30],
    'batch_2': [40, 50, 60],
    'batch_3': [70, 80, 90]
}

total_records = sum(len(v) for v in dataset.values())
print(f"✓ Cell 2 Complete: Dataset created")
print(f"Total records: {total_records}")
```

```python
# Cell 3: Process data
# This cell depends on Cell 2 (dataset exists)

processed_data = {}
for batch_name, values in dataset.items():
    processed_data[batch_name] = {
        'sum': sum(values),
        'avg': sum(values) / len(values),
        'count': len(values)
    }

print(f"✓ Cell 3 Complete: Data processed")
print(f"Processed batches:")
for batch, stats in processed_data.items():
    print(f"  {batch}: avg={stats['avg']:.1f}, count={stats['count']}")
```

## Part 2: Demonstrate Kernel State

Notice that variables defined earlier are still accessible. The kernel remembers them.

```python
# Cell 4: Verify state persistence
# This cell shows that variables from previous cells are still in memory

print("Current kernel state:")
print(f"  PROJECT_NAME = {PROJECT_NAME}")
print(f"  total_records = {total_records}")
print(f"  processed_data keys = {list(processed_data.keys())}")
print()
print("✓ All variables are accessible!")
print("This is because the kernel maintains state across cells.")
```

## Part 3: Interruption Demo

Run the cell below and then press the Stop button (⏹) in the toolbar to interrupt it.

**Instructions:**
1. Click the cell below
2. Press Ctrl+Enter to run it
3. After 2-3 seconds of output, press Ctrl+C or click the Stop button (⏹) in the toolbar
4. The cell will stop, but your kernelstate is preserved

```python
# Cell 5: Long-running operation (interrupt this!)
# Run this cell and then interrupt it with Ctrl+C

print("Starting long-running operation...")
print("Press Ctrl+C or click the Stop button to interrupt!")
print()

interrupt_count = 0
for i in range(100):
    interrupt_count += 1
    time.sleep(0.5)  # Wait 0.5 seconds each iteration
    print(f"Processing iteration {i+1}... (interrupt_count = {interrupt_count})")

print("Operation complete!")
```

```python
# Cell 6: Verify state after interruption
# After interrupting Cell 5, run this to show the kernel still has your variables

print("✓ Cell 6: Kernel is still responsive!")
print()
print("Notice:")
print(f"  - interrupt_count = {interrupt_count} (kernel state preserved!)")
print(f"  - PROJECT_NAME = {PROJECT_NAME} (still accessible)")
print()
print("This shows that interrupting a cell preserves kernel state.")
print("Only restarting the kernel clears everything.")
```

## Part 4: Restart Demo

Now perform a kernel restart:

**Instructions:**
1. Click on **Kernel** menu → **Restart Kernel**
2. After restart completes, run the cell below

```python
# Cell 7: Verify state after restart
# After restarting the kernel, run this cell

print("After kernel restart:")
print()

try:
    print(f"PROJECT_NAME = {PROJECT_NAME}")
except NameError:
    print("✗ PROJECT_NAME is not defined (cleared by restart)")

try:
    print(f"total_records = {total_records}")
except NameError:
    print("✗ total_records is not defined (cleared by restart)")

try:
    print(f"dataset = {dataset}")
except NameError:
    print("✗ dataset is not defined (cleared by restart)")

print()
print("All kernel memory has been cleared!")
```

## Part 5: Restore State After Restart

To get your variables back after a restart, simply run all cells from the beginning.

```python
# Cell 8: Run all cells (Ctrl+Click this or use Kernel → Run All Cells)
# 
# This cell shows the expected workflow:
# 1. Restart kernel (clears memory)
# 2. Run all cells from top (recreates variables)
# 3. Everything works again

print("")
print("=" * 50)
print("REPRODUCIBILITY CHECK")
print("=" * 50)
print()
print("After running all cells in order from Cell 1:")
print()

# Verify all expected variables exist
checks = [
    ('PROJECT_NAME', lambda: PROJECT_NAME == "Kernel Demo"),
    ('VERSION', lambda: VERSION == 1.0),
    ('dataset', lambda: len(dataset) == 3),
    ('total_records', lambda: total_records == 9),
    ('processed_data', lambda: len(processed_data) == 3)
]

all_passed = True
for var_name, check_func in checks:
    try:
        if check_func():
            print(f"✓ {var_name}: OK")
        else:
            print(f"✗ {var_name}: FAILED")
            all_passed = False
    except NameError:
        print(f"✗ {var_name}: NOT DEFINED")
        all_passed = False

print()
if all_passed:
    print("✓ ALL CHECKS PASSED!")
    print("This notebook is reproducible and works from scratch.")
else:
    print("✗ Some checks failed. Run cells in order from Cell 1.")
```

## Summary: Kernel Management Workflow

### Normal Workflow
1. ✓ Run cells in order (Cell 1 → Cell 2 → ... → Cell 8)
2. ✓ Verify outputs
3. ✓ If a cell is slow, **interrupt** it with Ctrl+C
4. ✓ If you need a fresh state, **restart** the kernel

### Key Insights
- **Kernel state persists** between cells until restart
- **Variables live in kernel memory**, not in the notebook file
- **Deleting a cell** doesn't delete its effects from kernel memory
- **Interrupting** stops a cell but keeps kernel state
- **Restarting** clears all memory and forces clean execution
- **Reproducibility** requires "Run All Cells" to work from scratch
