## 2. Data Preparation {#data-preparation}

### Step 1: Define the Dataset

We'll create a sample dataset containing student grades. This is not a real dataset, but it demonstrates how to structure and analyze education data.

The data structure includes:
- Student names as identifiers
- Three subject grades (Math, Science, English)
- Grade values ranging from 0 to 100

````python
# Step 1: Import required libraries
import pandas as pd
import statistics

print("✓ Libraries imported successfully")
````

````python
# Step 2: Create the student dataset
students_data = {
    'student_name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'math_grade': [92, 78, 85, 88, 95],
    'science_grade': [89, 82, 88, 91, 93],
    'english_grade': [85, 90, 80, 87, 89]
}

df = pd.DataFrame(students_data)
print("✓ Dataset created")
print(f"\nDataset shape: {df.shape} (rows, columns)")
print("\nFirst look at the data:")
print(df)
````

### Step 2: Verify Data Quality

Before analysis, we verify the dataset meets quality standards:
- All grades are within the valid range (0-100)
- No missing values exist
- Data types are correct

**Expected output**: Confirmation that all checks pass ✓

````python
# Data quality checks
print("Data Quality Verification:")
print("="*50)

# Check for missing values
missing_values = df.isnull().sum().sum()
print(f"Missing values: {missing_values}")
if missing_values == 0:
    print("✓ No missing data found\n")

# Check grade ranges
grade_columns = ['math_grade', 'science_grade', 'english_grade']
print("Grade Range Validation:")
for col in grade_columns:
    min_grade = df[col].min()
    max_grade = df[col].max()
    valid = (min_grade >= 0) and (max_grade <= 100)
    status = "✓" if valid else "✗"
    print(f"{status} {col}: {min_grade}-{max_grade}")

print("\n✓ All data quality checks passed!")
````

## 3. Analysis {#analysis}

### Key Questions

We answer these questions through analysis:
1. Which subject has the highest average grade?
2. Which student has the highest overall performance?
3. Are there any performance gaps between subjects?
4. How many students have grades above 85?

### Analysis Approach

Our analysis follows this sequence:
- Calculate subject averages
- Compute student overall grades
- Identify high performers
- Compare across subjects

### Example: Computing Averages

In Python, we calculate the average grade for each subject using the `pd.DataFrame.mean()` method:

```python
# Calculate average grade for each subject
subject_averages = df[['math_grade', 'science_grade', 'english_grade']].mean()

# Display results
for subject, average in subject_averages.items():
    print(f"{subject}: {average:.2f}")
```

This code:
1. Selects the three grade columns using list notation
2. Applies `.mean()` to calculate the average for each column
3. Loops through the results and prints them formatted to 2 decimal places

````python
# Calculate average grade for each subject
subject_averages = df[['math_grade', 'science_grade', 'english_grade']].mean()

print("Average Grade by Subject:")
print("="*40)
for subject, average in subject_averages.items():
    display_name = subject.replace('_', ' ').title()
    print(f"{display_name}: {average:.2f}")
````

### Student Overall Performance

We calculate each student's average grade across all three subjects.

**Methodology:**
- For each student, average their Math, Science, and English grades
- Store the overall average in a new column called `overall_grade`
- Sort students by performance (highest to lowest)

````python
# Calculate overall grade for each student
df['overall_grade'] = df[['math_grade', 'science_grade', 'english_grade']].mean(axis=1)

# Sort by overall grade (highest first)
df_sorted = df.sort_values('overall_grade', ascending=False)

print("Student Performance Rankings:")
print("="*60)
for idx, row in df_sorted.iterrows():
    name = row['student_name']
    overall = row['overall_grade']
    print(f"{name:12} - Overall Grade: {overall:.2f}")
````

### High Performers Identification

We identify students who achieve grades above 85 in their overall performance.

**Significance**: A grade above 85 typically indicates excellent performance and mastery of subject material.

````python
# Identify high performers (overall grade > 85)
high_performers = df[df['overall_grade'] > 85]
high_performers_count = len(high_performers)

print(f"High Performers (Grade > 85): {high_performers_count} students")
print("="*50)
for idx, row in high_performers.iterrows():
    name = row['student_name']
    overall = row['overall_grade']
    print(f"✓ {name}: {overall:.2f}")
````

## 4. Results {#results}

### Summary of Findings

**Subject Performance Summary:**

````python
# Create summary statistics
print("Summary Statistics:")
print("="*60)

# Subject averages
print("\n1. Average Grade by Subject (Higher is Better):\n")
subject_stats = df[['math_grade', 'science_grade', 'english_grade']].describe().loc[['mean', 'min', 'max']]
subject_stats.columns = ['Math', 'Science', 'English']
print(subject_stats.round(2))

# Overall statistics
print(f"\n2. Overall Class Performance:\n")
print(f"   Highest Overall Grade: {df['overall_grade'].max():.2f}")
print(f"   Lowest Overall Grade:  {df['overall_grade'].min():.2f}")
print(f"   Class Average:         {df['overall_grade'].mean():.2f}")

print("\n✓ Analysis complete!")
````

### Key Insights

Based on our analysis, we observe:

1. **Strong Overall Performance**
   - The class average exceeds 85, indicating strong student achievement
   - Most students demonstrate consistent performance across subjects

2. **Subject Comparison**
   - Science has the highest average, suggesting strong STEM aptitude
   - English grades remain solid but show slight variation

3. **Top Performers**
   - Several students achieve grades above 90
   - High performers show balanced excellence across all subjects
