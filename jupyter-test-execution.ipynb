## 4. Data Science Libraries Check

Verify that key data science packages are installed.

```python
# Check for common data science libraries
libraries = {
    'numpy': 'NumPy',
    'pandas': 'Pandas',
    'matplotlib': 'Matplotlib',
    'sklearn': 'Scikit-Learn'
}

print("Data Science Libraries Status:")
print("-" * 50)

for module_name, display_name in libraries.items():
    try:
        module = __import__(module_name)
        version = getattr(module, '__version__', 'unknown')
        print(f"✅ {display_name}: {version} - INSTALLED")
    except ImportError:
        print(f"❌ {display_name}: NOT INSTALLED")
```

## 5. Simple Data Analysis Example

Verify that Jupyter can execute real-world data science code.

```python
# Create a simple dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 92, 78],
    'Status': ['Pass', 'Pass', 'Pass']
}

# Try importing pandas (data analysis library)
try:
    import pandas as pd
    df = pd.DataFrame(data)
    print("\nDataFrame created successfully:")
    print(df)
    print(f"\nAverage Score: {df['Score'].mean():.2f}")
except ImportError:
    print("Pandas not installed, but basic Python operations work:")
    for i, name in enumerate(data['Name']):
        print(f"{name}: {data['Score'][i]}")
```
